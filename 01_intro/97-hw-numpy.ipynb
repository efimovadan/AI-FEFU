13) Вектор чисел от 1 до 51 преобразовать в вектор нечетных чисел и в вектор четных чисел, поменять их размерности, чтобы получились две матрица. Найти произведение этих матриц. Найти обратную матрицу эётого произведения.  

```python
vec = np.arange(1, 51)

odd_numbers = vec[vec % 2 == 1] 
even_numbers = vec[vec % 2 == 0] 

matrix_odd = odd_numbers.reshape(5, 5)
matrix_even = even_numbers.reshape(5, 5)

product_matrix = np.dot(matrix_odd, matrix_even)

try:
    inverse_matrix = np.linalg.inv(product_matrix)
    print("Обратная матрица произведения:\n", inverse_matrix)
except np.linalg.LinAlgError:
    print("Произведение матриц вырождено, обратная матрица не существует.")
```
